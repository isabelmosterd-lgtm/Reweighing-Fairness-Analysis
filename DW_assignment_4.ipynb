```python
import pandas as pd
import kagglehub
import os
from ucimlrepo import fetch_ucirepo

# 1. Adult Census Income (Kaggle)
adult_path = kagglehub.dataset_download("uciml/adult-census-income")
print("Adult files:", os.listdir(adult_path))

# 2. German Credit (UCI)
german_data = fetch_ucirepo(id=144)
german_X = german_data.data.features
german_y = german_data.data.targets

# 3. COMPAS (Kaggle)
compas_path = kagglehub.dataset_download("danofer/compass")
print("COMPAS files:", os.listdir(compas_path))

# Check dataset dimensions
print("German features:", german_X.shape)
print("German target:", german_y.shape)
```

```python

# Load Adult
adult = pd.read_csv(os.path.join(adult_path, "adult.csv"))

# Load German Credit
german = german_X.copy()
german["target"] = german_y.iloc[:, 0].values

# Load COMPAS (violent recidivism version)
compas = pd.read_csv(
    os.path.join(compas_path, "cox-violent-parsed_filt.csv")
)

# Check dimensions
print("Adult:", adult.shape)
print("German:", german.shape)
print("COMPAS:", compas.shape)

# Preview datasets
display(adult.head())
display(german.head())
display(compas.head())
```

```python

# Adult: create binary target and sensitive attribute
adult["y"] = (adult["income"] == ">50K").astype(int)
adult["s"] = (adult["sex"] == "Male").astype(int)

print(adult[["income", "sex", "y", "s"]].head())
```

```python

# German Credit: binary target
german["y"] = (german["target"] == 1).astype(int)

# Sensitive attribute: 1 = Male, 0 = Female
german["s"] = german["Attribute9"].map({
    "A91": 1,
    "A92": 0,
    "A93": 1,
    "A94": 1
})

print(german[["Attribute9", "target", "y", "s"]].head())
```

```python

# COMPAS: select the two groups
compas_clean = compas[
    compas["race"].isin(["African-American", "Caucasian"])
    & compas["is_recid"].isin([0, 1])
].copy()

# Positive outcome: no recidivism
compas_clean["y"] = (compas_clean["is_recid"] == 0).astype(int)

# Sensitive attribute: 1 = Caucasian, 0 = African-American
compas_clean["s"] = (
    compas_clean["race"] == "Caucasian"
).astype(int)

print(compas_clean[["race", "is_recid", "y", "s"]].head())
```

```python

# Calculate fairness metrics on the original labels

def fairness_metrics(df):
    privileged_rate = df.loc[df["s"] == 1, "y"].mean()
    unprivileged_rate = df.loc[df["s"] == 0, "y"].mean()

    dp_diff = privileged_rate - unprivileged_rate

    di_ratio = (
        unprivileged_rate / privileged_rate
        if privileged_rate > 0 else float("nan")
    )

    return {
        "Privileged positive rate": privileged_rate,
        "Unprivileged positive rate": unprivileged_rate,
        "DP Difference": dp_diff,
        "DI Ratio": di_ratio
    }

results = pd.DataFrame({
    "Adult": fairness_metrics(adult),
    "German": fairness_metrics(german),
    "COMPAS": fairness_metrics(compas_clean)
}).T

display(results.round(3))
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

def prepare_features(df, exclude_columns):
    X = df.drop(columns=exclude_columns, errors="ignore").copy()

    # Replace question marks with missing values
    X = X.replace("?", np.nan)

    numeric_cols = X.select_dtypes(include="number").columns.tolist()
    categorical_cols = X.select_dtypes(exclude="number").columns.tolist()

    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer([
        ("num", numeric_pipeline, numeric_cols),
        ("cat", categorical_pipeline, categorical_cols)
    ])

    return preprocessor.fit_transform(X)
```

```python

# Select features for Adult consistency analysis
adult_features = [
    "age",
    "workclass",
    "education.num",
    "marital.status",
    "occupation",
    "relationship",
    "capital.gain",
    "capital.loss",
    "hours.per.week"
]

X_adult = adult[adult_features].copy()
y_adult = adult["y"]

print("Features:", X_adult.shape)
print("Labels:", y_adult.shape)
display(X_adult.head())
```

```python

from sklearn.neighbors import NearestNeighbors
import numpy as np

# Prepare numerical features
X_adult_prepared = prepare_features(
    X_adult,
    exclude_columns=[]
)

# Calculate consistency using 5 nearest neighbors
def calculate_consistency(X, y, k=5):
    neighbors = NearestNeighbors(
        n_neighbors=k + 1,
        algorithm="brute"
    )
    neighbors.fit(X)

    # Include one extra neighbor because each row is its own neighbor
    indices = neighbors.kneighbors(
        X, return_distance=False
    )

    # Exclude each observation itself
    neighbor_indices = indices[:, 1:k+1]

    y_values = np.asarray(y)
    neighbor_mean = y_values[neighbor_indices].mean(axis=1)

    consistency = 1 - np.mean(
        np.abs(y_values - neighbor_mean)
    )

    return consistency

adult_consistency = calculate_consistency(
    X_adult_prepared, y_adult
)

print("Adult consistency:", round(adult_consistency, 3))
```

```python

# Select features for German Credit consistency
# Exclude target and sensitive attribute

german_features = [
    col for col in german.columns
    if col not in ["target", "y", "s", "Attribute9"]
]

X_german = german[german_features].copy()
y_german = german["y"]

# Prepare features
X_german_prepared = prepare_features(
    X_german,
    exclude_columns=[]
)

# Calculate consistency
german_consistency = calculate_consistency(
    X_german_prepared,
    y_german
)

print("German consistency:", round(german_consistency, 3))
```

```python

from sklearn.model_selection import train_test_split

# Split Adult into training (80%) and test (20%)
X_train, X_test, y_train, y_test, s_train, s_test = train_test_split(
    X_adult,
    y_adult,
    adult["s"],
    test_size=0.2,
    random_state=42,
    stratify=y_adult
)

print("Training samples:", len(X_train))
print("Test samples:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTest target distribution:")
print(y_test.value_counts(normalize=True))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline

# Create baseline classifier
baseline_model = Pipeline([
    ("preprocessing", ColumnTransformer([
        ("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]), X_train.select_dtypes(include="number").columns.tolist()),

        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), X_train.select_dtypes(exclude="number").columns.tolist())
    ])),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train without Reweighing
baseline_model.fit(X_train, y_train)

# Predict on untouched test data
y_pred_baseline = baseline_model.predict(X_test)

# Calculate accuracy
baseline_accuracy = accuracy_score(y_test, y_pred_baseline)

print("Baseline accuracy:", round(baseline_accuracy, 3))
```

```python

# Calculate fairness of baseline predictions

baseline_fairness = fairness_metrics(
    pd.DataFrame({
        "y": y_pred_baseline,
        "s": s_test.to_numpy()
    })
)

print("Baseline accuracy:", round(baseline_accuracy, 3))
print("Baseline DP Difference:", round(baseline_fairness["DP Difference"], 3))
print("Baseline DI Ratio:", round(baseline_fairness["DI Ratio"], 3))
```

```python

# Calculate Reweighing weights using training data only

train_groups = pd.DataFrame({
    "s": s_train.to_numpy(),
    "y": y_train.to_numpy()
})

n = len(train_groups)

group_counts = train_groups["s"].value_counts()
outcome_counts = train_groups["y"].value_counts()
joint_counts = train_groups.groupby(["s", "y"]).size()

# Calculate a weight for each training example
weights = []

for s, y in zip(train_groups["s"], train_groups["y"]):
    weight = (
        group_counts[s] * outcome_counts[y]
        / (n * joint_counts[s, y])
    )
    weights.append(weight)

# Show weights for the four combinations
train_groups["weight"] = weights

print(
    train_groups.groupby(["s", "y"])["weight"]
    .first()
    .round(3)
)
```

```python

from sklearn.base import clone

# Create a new model with the same settings
reweighing_model = clone(baseline_model)

# Train using Reweighing weights
reweighing_model.fit(
    X_train,
    y_train,
    classifier__sample_weight=weights
)

# Predict on the same test set
y_pred_reweighing = reweighing_model.predict(X_test)

# Calculate accuracy
reweighing_accuracy = accuracy_score(
    y_test,
    y_pred_reweighing
)

# Calculate fairness
reweighing_fairness = fairness_metrics(
    pd.DataFrame({
        "y": y_pred_reweighing,
        "s": s_test.to_numpy()
    })
)

print("Reweighing accuracy:", round(reweighing_accuracy, 3))
print("Reweighing DP Difference:", round(reweighing_fairness["DP Difference"], 3))
print("Reweighing DI Ratio:", round(reweighing_fairness["DI Ratio"], 3))
```

```python

from sklearn.neighbors import NearestNeighbors

def calculate_consistency(X, y, k=5):
    neighbors = NearestNeighbors(
        n_neighbors=k,
        algorithm="brute"
    )
    neighbors.fit(X)

    # Exclude each sample itself from its neighbors
    neighbor_indices = neighbors.kneighbors(
        return_distance=False
    )

    y_values = np.asarray(y)
    neighbor_mean = y_values[neighbor_indices].mean(axis=1)

    return 1 - np.mean(np.abs(y_values - neighbor_mean))

# Use the fitted training preprocessor
preprocessor = baseline_model.named_steps["preprocessing"]

# Transform test features without fitting again
X_test_prepared = preprocessor.transform(X_test)

# Calculate consistency of both models
baseline_consistency = calculate_consistency(
    X_test_prepared,
    y_pred_baseline
)

reweighing_consistency = calculate_consistency(
    X_test_prepared,
    y_pred_reweighing
)

print("Baseline consistency:", round(baseline_consistency, 3))
print("Reweighing consistency:", round(reweighing_consistency, 3))
```
